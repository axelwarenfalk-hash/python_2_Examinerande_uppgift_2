```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
api_url_names = "https://motenmedborgarportal.malmo.se/api/err/v1.0/representatives?committeeId=909&pageSize=9999"
api_response_names = requests.get(api_url_names)

print(api_response_names.status_code)
politician_names = api_response_names.json()

politician_names
```

```python
politician_roles_df = pd.json_normalize(
    politician_roles,
    sep="_"
)


politician_roles_df = politician_roles_df[
    [
        "RepresentativeId",
        "Role",
        "DateFrom",
        "DateTo",
        "Number",
        "State",
        "Committee_Name"
    ]
]

display(politician_roles_df.head())
print(len(politician_roles_df))
```

```python
all_lines = []
with pdfplumber.open(BytesIO(response.content)) as pdf:
    for page in pdf.pages:
        text = page.extract_text() or ""
        all_lines.extend(text.splitlines())
```

```python
print(all_lines)
```

```python
voting_members = []

voting_member_section = False

for line in all_lines:
    lower = line.lower()

    if lower.strip().isdigit():
        continue

    if "beslutande ledamöter" in lower:
        voting_member_section = True
        continue

    if "ersätter" in lower:
        break

    if voting_member_section:
        voting_members.append(line)

    
```

```python
def split_first_member(member: str) -> tuple[dict[str, str | None], str]:

    name, rest = member.strip().split("(", 1)
    name = name.strip()

    party, rest = rest.strip().split(")",1)
    party = party.strip()

    rest = rest.strip()

    if rest.startswith('('):
        title, rest = rest.split(')',1)
        title = title.replace('(', '')
    else:
        title = None

    structured_member = {
        'name': name,
        'party': party,
        'title': title
    }

    return structured_member, rest.strip()
```

```python
def paragraph_interval(paragraph: str) -> list[int] | None:
    if '§' not in paragraph:
        return None

    paragraph = paragraph.replace('§', '').strip()
    paragraphs = paragraph.split(',')

    expanded = []

    for part in paragraphs:
        part = part.strip()

        if '-' in part:
            start, end = map(int, part.split('-'))
            expanded.extend(range(start, end + 1))
        else:
            expanded.append(int(part))

    return expanded
```

```python
voting_members_dict = []

for member in voting_members:

    structured_member, rest = split_first_member(member)
    paragraph = paragraph_interval(rest)

    voting_members_dict.append(
        {
            'name': structured_member['name'],
            'party': structured_member['party'],
            'paragraphs': paragraph
        }
    )


print(voting_members_dict)
```

```python
voting_extras = []

voting_member_section = False

for line in all_lines:
    lower = line.lower()

    if voting_members[-1].lower() in lower:
        voting_member_section = True
        continue

    if "ej tjänstgörande ersättare" in lower:
        break

    if voting_member_section:
        voting_extras.append(line)
```

```python
print(voting_extras)
```

```python
import re
```

```python
voting_extras_merged = []

extra_row_number = -1

for i, extra in enumerate(voting_extras):
    extra = extra.strip()

    # alltid 2 §§, alltid ett intevall på max 3 tecken upp till 999. alltid mellanslag 'ersätter'. alltid parti inom parates i slutet.
    correct_syntax = r"^§§\d{1,3}(?:-\d{1,3})?\s+ersätter\s+.+?\s+\([A-Za-zÅÄÖåäö]+\)\s*$"

    if extra_row_number == i:
        continue

    if ',' in extra:
        after_comma = extra.split(',',1)[1].strip()

        if re.match(correct_syntax, after_comma):
            voting_extras_merged.append(extra)
        else:
            merged = extra + ' ' + voting_extras[i+1]
            voting_extras_merged.append(merged)
            extra_row_number = i + 1

    else:
        voting_extras_merged.append(extra)

print(voting_extras_merged)
```

```python
voting_extras_dict = []

for extra in voting_extras_merged:
    extra, rest = split_first_member(extra)
    extra['replaces'] = []

    rest = rest.split(',')

    for member in rest:
        member = member.split('ersätter')

        paragraph = member[0].strip()
        paragraphs = paragraph_interval(paragraph)

        name = member[1].strip()
        name, rest = split_first_member(name)

        extra['replaces'].append(
            {
                'member': name,
                'paragraphs': paragraphs
            }
        )

    voting_extras_dict.append(extra)

print(voting_extras_dict)
```
